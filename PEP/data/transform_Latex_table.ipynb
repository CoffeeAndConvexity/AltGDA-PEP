```python
import numpy as np
import pandas as pd
```

```python
AltGDA_data = pd.read_csv('./AltGDA.csv')
```

```python
# transform the data into the form of "$5$ & $1.523$ & $0.614$ \\"

combined = []
for index, row in AltGDA_data.iterrows():
    output_line = f"${int(row['N'])}$ & ${row['optimal_η']:.3f}$ & ${row['optimal_obj']:.3f}$ \\\\"
    combined.append(output_line)

print('\n'.join(combined))
```
